```python
import pandas as pd
df=pd.read_csv('breast_cancer_clinical1.csv')
import scipy
from scipy.stats import shapiro
stat,p=shapiro(df['Age'])
print('age:',stat,p)
import matplotlib.pyplot as plt
plt.hist(df['Age'],bins=10)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
import scipy.stats as stats
import matplotlib.pyplot as plt
stats.probplot(df['Age'],dist='norm',plot=plt)
plt.show()
group0=df[df['Lymph_Node']==0]
group1=df[df['Lymph_Node']==1]
import scipy
from scipy.stats import shapiro
stat0,p0=shapiro(group0['Age'])
stat1,p1=shapiro(group1['Age'])
print('Lymph_Node0:',stat0,p0)
print('Lymph_Node1:',stat1,p1)
import matplotlib.pyplot as plt
plt.hist(group0['Age'],bins=10)
plt.xlabel('LN0age')
plt.ylabel('Frequency')
plt.show()
import scipy.stats as stats
import matplotlib.pyplot as plt
stats.probplot(group0['Age'],dist='norm',plot=plt)
plt.show()
plt.hist(group1['Age'],bins=10)
plt.xlabel('LN1age')
plt.ylabel('Frequency')
plt.show()
stats.probplot(group1['Age'],dist='norm',plot=plt)
plt.show()
stat,p=shapiro(df['Tumor_Size'])
print('----------')
print('tumor_size:',stat,p)
group0=df[df['Lymph_Node']==0]
group1=df[df['Lymph_Node']==1]
stat0,p0=shapiro(group0['Tumor_Size'])
print('Lymph_Node0:',stat0,p0)
stat1,p1=shapiro(group1['Tumor_Size'])
print('Lymph_Node1:',stat1,p1)
print('----------')
stat,p=shapiro(df['Ki67'])
print('Ki67:',stat,p)
group0=df[df['Lymph_Node']==0]
group1=df[df['Lymph_Node']==1]
stat0,p0=shapiro(group0['Ki67'])
stat1,p1=shapiro(group1['Ki67'])
print('LN0ki67:',stat0,p0)
print('LN1ki67',stat1,p1)
print('----------')
df=pd.DataFrame({
    'Variable':['Age','Age','Age','Tumor_Size','Tumor_Size','Tumor_Size','Ki67','Ki67','Ki67'],
    'Group':['Overall','LN(-)','LN(+)','Overall','LN(-)','LN(+)','Overall','LN(-)','LN(+)'],
    'P value':['0.98','0.75','0.88','0.92','0.97','0.85','0.61','0.80','0.99'],
    'Distribution':['Normal','Normal','Normal','Normal','Normal','Normal','Normal','Normal','Normal'],
    'Next analysis':['t-test','t-test','t-test','t-test','t-test','t-test','t-test','t-test','t-test']
})
print(df)
df.to_excel('Normality Test Results.xlsx')
```

```python
import pandas as pd
df=pd.read_csv('breast_cancer_clinical3.csv')
print('breast_cancer_clinical3读取成功')
print('-----------')
print('Univariate Logistic Regression Analysis:')
import statsmodels
import statsmodels.api as sm
import numpy as np
variables=[
    'Age',
    'Tumor_Size',
    'Ki67',
    'ER',
    'PR',
    'HER2'
]
results=[]
for variable in variables:
    y=df['Lymph_Node']
    X=df[[variable]]
    X=sm.add_constant(X)
    model=sm.Logit(y,X).fit(disp=False)
    beta=model.params[variable]
    p=model.pvalues[variable]
    OR=np.exp(model.params[variable])
    CI=np.exp(model.conf_int().loc[variable])
    results.append([
        variable,
        beta,
        OR,
        CI[0],
        CI[1],
        p
    ])
table1=pd.DataFrame(
    results,
    columns=[
        'Variable',
        'Beta',
        'OR',
        '95%CI上限',
        '95%CI下限',
        'P'
    ])
print(table1)
table1.to_excel('Univariate Logistic Regression Analysis.xlsx')
print('Univariate Logistic Regression Analysis.xlsx保存成功')
print('----------')
print('multivariate Logistic Regression Analysis:')
y=df['Lymph_Node']
X=df[['Age','Tumor_Size','Ki67','ER','PR','HER2']]
X=sm.add_constant(X)
model=sm.Logit(y,X).fit(disp=False)
print('Beta:')
beta=model.params
print(beta)
p=model.pvalues
print('P')
print(p)
OR=np.exp(model.params)
print('OR:')
print(OR)
CI=np.exp(model.conf_int())
print(CI[0])
print('CI[1]:')
print(CI[1])
table2=pd.DataFrame({
    'Beta':beta,
    'OR':OR,
    '95%CI上限':CI[0],
    '95%CI下限':CI[1],
    'P':p
})
print('方式一table2:')
print(table2)
print('单因素Logistic回归分析显示，年龄、肿瘤大小、PR、HER2均与淋巴结转移有统计学关联。')
table2.to_excel('multivariate Logistic Regression Analysis.xlsx')
print('multivariate Logistic Regression Analysis.xlsx保存成功')
results1=[]
for variable in variables:
    beta=model.params[variable]
    OR=np.exp(model.params[variable])
    CI=np.exp(model.conf_int().loc[variable])
    p=model.pvalues[variable]
    results1.append([
        variable,
        beta,
        OR,
        CI[0],
        CI[1],
        p
    ])
table3=pd.DataFrame(
    results1,
    columns=[
        'Variable',
        'Beta',
        'OR',
        '95%CI上限',
        '95%CI下限',
        'P'
    ])
print('方式二table3:')
print(table3)
print('多因素Logistic回归分析结果分析显示，P值都大于0.05，因此都不是独立相关因素')
        
    
```

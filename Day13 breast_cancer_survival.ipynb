```python
import pandas as pd
import numpy as np
np.random.seed(42)
df=pd.DataFrame({
    'Patient_ID':range(1,101),
    'Age':np.random.randint(35,80,100),
    'Tumor_size':np.random.uniform(1,8,100).round(2),
    'TNM_stage':np.random.choice(['I','II','III'],100),
    'ER':np.random.choice(['-','+'],100),
    'PR':np.random.choice(['-','+'],100),
    'HER2':np.random.choice(['-','+'],100),
    'Ki67':np.random.randint(5,80,100),
    'Survival_months':np.random.randint(10,120,100),
    'Status':np.random.choice([0,1],100)
})
print('数据结构：')
print(df)
print(df.info())
print(df.describe())
df.to_csv('breast_cancer_survival.csv')
print('breast_cancer_survival.csv保存成功')
print('------------')
print('总生存曲线:')
from lifelines import KaplanMeierFitter
import matplotlib.pyplot as plt
kmf=KaplanMeierFitter()
kmf.fit(
    df['Survival_months'],
    df['Status']
)
kmf.plot()
plt.title('Overall survival curve')
plt.xlabel('Time(months)')
plt.ylabel('Survival probability')
plt.savefig(
    'Overall_survival_curve.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
print('------------')
print('TNM分期KM曲线:')
stages=['I','II','III']
for stage in stages:
    stage_data=df[df['TNM_stage']==stage]
    kmf.fit(
        stage_data['Survival_months'],
        stage_data['Status'],
        label='Stage'+stage
    )
    kmf.plot()
plt.title('Survival curves by stage')
plt.xlabel('Time(months)')
plt.ylabel('Survival probability')
plt.legend()
plt.savefig(
    'Survival curves by stage.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
print('TNM分期KM曲线的Log-rank分析：')
from lifelines.statistics import multivariate_logrank_test
result_stage=multivariate_logrank_test(
    df['Survival_months'],
    df['Status'],
    df['TNM_stage']
)
print('TNM分期的KM曲线p值：',result_stage.p_value)
print('因为p值>0.05,所以TNM分期的生存曲线没有统计学意义。')
print('----------')
print('ER的KM曲线：')
for group_er in ['-','+']:
    ix=df['ER']==group_er
    kmf.fit(
        df.loc[ix,'Survival_months'],
        df.loc[ix,'Status'],
        label='ER'+group_er
    )
    kmf.plot()
plt.title('Survival curves by ER')
plt.xlabel('Time(months)')
plt.ylabel('Survival probability')
plt.legend()
plt.savefig(
    'Survival curves by ER.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
print('ER的KM曲线的Log-rank检验：')
from lifelines.statistics import logrank_test
result_ER=logrank_test(
    df['Survival_months'],
    df['Status'],
    df['ER']
)
print('ER的KM曲线p值：',result_ER.p_value)
print('因为P值<0.05，所以ER的KM曲线有统计学意义。')
print('----------')
print('PR的KM曲线：')
for group_pr in ['-','+']:
    ix=df['PR']==group_pr
    kmf.fit(
        df.loc[ix,'Survival_months'],
        df.loc[ix,'Status'],
        label='PR'+group_pr
    )
    kmf.plot()
plt.title('Survival curves by PR')
plt.xlabel('Time(months)')
plt.ylabel('Survival probability')
plt.legend()
plt.savefig(
    'Survival curves by PR.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
print('PR的KM曲线的Log-rank检验：')
result_PR=logrank_test(
    df['Survival_months'],
    df['Status'],
    df['PR']
)
print('PR的KM曲线P值：',result_PR.p_value)
print('因为p值<0.05,所以PR的KM曲线有统计学意义。')
print('-----------')
print('HER2的KM曲线：')
for group_her2 in ['-','+']:
    ix=df['HER2']==group_her2
    kmf.fit(
        df.loc[ix,'Survival_months'],
        df.loc[ix,'Status'],
        label='HER2'+group_her2
    )
    kmf.plot()
plt.title('Survival curves by HER2')
plt.xlabel('Time(months)')
plt.ylabel('Survival probability')
plt.legend()
plt.savefig(
    'Survival curves by HER2.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
print('HER2的KM曲线的Log-rank检验：')
result_HER2=logrank_test(
    df['Survival_months'],
    df['Status'],
    df['HER2']
)
print('HER2的KM曲线p值：',result_HER2.p_value)
print('因为p<0.05,所以HER2的KM曲线有统计学意义。')
print('------------')
print('Ki67的KM曲线：')
df['Ki67_group']=df['Ki67'].apply(lambda x:'high' if x>=30 else 'low')
for group_ki67 in ['low','high']:
    ix=df['Ki67_group']==group_ki67
    kmf.fit(
        df.loc[ix,'Survival_months'],
        df.loc[ix,'Status'],
        label='Ki67'+group_ki67
    )
    kmf.plot()
plt.title('Survival curves by Ki67')
plt.xlabel('Time(months)')
plt.ylabel('Survival probability')
plt.legend()
plt.savefig(
    'Survival curves by Ki67.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
print('Ki67的KM曲线Log-rank检验：')
result_Ki67=logrank_test(
    df['Survival_months'],
    df['Status'],
    df['Ki67_group']
)
print('Ki67的KM曲线p值：',result_Ki67.p_value)
print('因为p值<0.05，所以Ki67的KM曲线有统计学意义。')
print('-----------')
print('单因素Cox回归：')
from lifelines import CoxPHFitter
cph=CoxPHFitter()
cph.fit(
    df[['Survival_months','Status','Age']],
    duration_col='Survival_months',
    event_col='Status'
)
print('Age:')
cph.print_summary()
print('C-index:',cph.concordance_index_)
print('因为C-index=0.49，故Age的Cox回归模型属于完全随机预测。且p>0.05,ER对生存结局之间没有统计学关联。')
print('-----------')
print('ER:')
df['ER_binary']=(df['ER']=='+').astype(int)
cph.fit(
    df[['Survival_months','Status','ER_binary']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('C-index:',cph.concordance_index_)

print('--------------')
print('PR:')
df['PR_binary']=(df['PR']=='+').astype(int)
cph.fit(
    df[['Survival_months','Status','PR_binary']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('C-index:',cph.concordance_index_)
print('因为C-index=0.48，故PR的Cox回归模型属于完全随机预测。且p>0.05,PR对生存结局之间没有统计学关联。')
print('-----------')
print('TNM_stage:')
df['Stage_binary']=(df['TNM_stage']=='III').astype(int)
cph.fit(
    df[['Survival_months','Status','Stage_binary']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('C-index:',cph.concordance_index_)
print('因为C-index=0.56，故TNM-stage的Cox回归模型属于完全随机预测。且p>0.05,TNM分期对生存结局之间没有统计学关联。')
print('-----------')
print('Tumor_size:')
cph.fit(
    df[['Survival_months','Status','Tumor_size']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('C-index:',cph.concordance_index_)
print('因为C-index=0.48，故Tumor_size的Cox回归模型属于完全随机预测。且p>0.05,肿瘤大小对生存结局之间没有统计学关联。')
print('-----------')
print('HER2:')
df['HER2_binary']=(df['HER2']=='+').astype(int)
cph.fit(
    df[['Survival_months','Status','HER2_binary']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('C-index:',cph.concordance_index_)
print('因为C-index=0.49，故HER2的Cox回归模型属于完全随机预测。且p>0.05,HER2对生存结局之间没有统计学关联。')
print('------------')
print('Ki67:')
df['Ki67_binary']=(df['Ki67']>=30).astype(int)
cph.fit(
    df[['Survival_months','Status','Ki67_binary']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('C-index:',cph.concordance_index_)
print('因为C-index=0.53，故Ki67的Cox回归模型属于完全随机预测。且p>0.05,Ki67对生存结局之间没有统计学关联。')
print('-----------')
print('因为所有可能影响生存的变量的P值均大于0.05，所以这些都不是影响生存结局的相关变量，无需进行下一步的多因素Cox。')
print('假设Age、Tumor_size、TNM_stage是单因素显著变量，进行下一步的多因素Cox分析，以筛选出独立危险因素。')
print('------------')
print('多因素回归分析：')
cph.fit(
    df[['Survival_months','Status','Age','Tumor_size','Stage_binary']],
    duration_col='Survival_months',
    event_col='Status'
)
cph.print_summary()
print('因为Age、Tumor_size、TNM_stage的P值均大于0.05，所以均没有统计学意义。')
print('因为C-index=0.59,所以Age、Tumor_size、TNM_stage的多因素Cox回归分析模型属于随机预测模型。')
```

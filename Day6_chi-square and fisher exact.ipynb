```python
import pandas as pd
df=pd.read_csv('breast_cancer_clinical3.csv')
table1=pd.crosstab(df['ER'],df['Lymph_Node'])
table2=pd.crosstab(df['PR'],df['Lymph_Node'])
table3=pd.crosstab(df['HER2'],df['Lymph_Node'])
print('ER与LN的交叉表：')
print(table1)
print('----------')
print('PR与LN的交叉表：')
print(table2)
print('----------')
print('HER2与LN的交叉表')
print(table3)
print('----------')
import scipy
from scipy.stats import chi2_contingency
chi2,p,dof,expected=chi2_contingency(table1)
print('ER的chi2:',chi2)
print('ER的p值：',p)
print('ER的dof:',dof)
print('ER的expected:',expected)
print('---------')
chi2,p,dof,expected=chi2_contingency(table2)
print('PR的chi2:',chi2)
print('PR的p值:',p)
print('PR的dof:',dof)
print('PR的expected:',expected)
print('----------')
chi2,p,dof,expected=chi2_contingency(table3)
print('HER2的chi2:',chi2)
print('HER2的p值:',p)
print('HER2的dof:',dof)
print('HER2的expected:',expected)
print('---------')
print('chi square table:')
table4=pd.DataFrame({
    'variable':['ERvsLN','PRvsLN','HER2vsLN'],
    'chi2':[2.42,4.9,3.23],
    'p':[0.12,0.027,0.072]
})
print(table4)
print('PR的表达状态与淋巴结转移存在统计学关联')
print('ER和HER2的表达状态与淋巴结转移未发现显著关联')
table4.to_excel('chi2 analysis table.xlsx')
print('-----------')
from scipy.stats import fisher_exact
oddsratio,p=fisher_exact(table1)
print('ER与LN的OR:',oddsratio)
print('ER与LN的p值:',p)
print('----------')
oddsratio,p=fisher_exact(table2)
print('PR与LN的OR:',oddsratio)
print('PR与LN的p值:',p)
print('----------')
oddsratio,p=fisher_exact(table3)
print('HER2与LN的OR：',oddsratio)
print('HER2与LN的p值:',p)
print('----------')
table5=pd.DataFrame({
    'variable':['ERvsLN','PRvsLN','HER2vsLN'],
    'OR':[0.1,0.63,9.33],
    'p':[0.07,0.02,0.07]
})
print('fisher_exact table:')
print(table5)
print('PR的表达状态与淋巴结转移存在统计学关联')
print('ER与HER2的表达状态与淋巴结转移未发现明显关联')
table5.to_excel('fisher_exact_table.xlsx')
```
